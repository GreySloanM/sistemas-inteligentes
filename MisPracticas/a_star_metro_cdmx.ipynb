# Algoritmo A* aplicado al Metro de la CDMX

En esta práctica se implementará el algoritmo de búsqueda informada A* para encontrar una ruta entre estaciones del Metro de la Ciudad de México.

El algoritmo A* utiliza la función:

\[
f(n) = g(n) + h(n)
\]

donde:

- \(g(n)\) es el costo acumulado desde la estación inicial.
- \(h(n)\) es una estimación de la distancia desde la estación actual hasta la estación destino.
- \(f(n)\) representa el costo estimado total de la ruta.

```python
import heapq
```

```python
class Problem:
    def __init__(self, initial, goal=None):
        self.initial = initial
        self.goal = goal

    def actions(self, state):
        raise NotImplementedError

    def result(self, state, action):
        raise NotImplementedError

    def is_goal(self, state):
        return state == self.goal

    def action_cost(self, state1, action, state2):
        return 1
```

```python
class GraphAStarProblem(Problem):
    def __init__(self, initial, goal, graph, heuristic):
        super().__init__(initial, goal)
        self.graph = graph
        self.heuristic = heuristic

    def actions(self, state):
        return list(self.graph[state].keys())

    def result(self, state, action):
        return action

    def action_cost(self, state1, action, state2):
        return self.graph[state1][state2]

    def h(self, state):
        return self.heuristic[state]
```

```python
class Node:
    def __init__(self, state, parent=None, action=None, path_cost=0):
        self.state = state
        self.parent = parent
        self.action = action
        self.path_cost = path_cost

    def path(self):
        lista_path = []
        node = self

        while node:
            lista_path.append(node.state)
            node = node.parent

        return lista_path[::-1]

    def expand(self, problem):
        lista = []

        for action in problem.actions(self.state):
            lista.append(self.child_node(problem, action))

        return lista

    def child_node(self, problem, action):
        next_state = problem.result(self.state, action)

        step_cost = problem.action_cost(
            self.state,
            action,
            next_state
        )

        return Node(
            next_state,
            self,
            action,
            self.path_cost + step_cost
        )
```

```python
def calcular_heuristica(grafo, destino):
    distancias = {estacion: float("inf") for estacion in grafo}
    distancias[destino] = 0

    frontera = [(0, destino)]

    while frontera:
        distancia_actual, estacion_actual = heapq.heappop(frontera)

        if distancia_actual > distancias[estacion_actual]:
            continue

        for vecino, distancia in grafo[estacion_actual].items():

            nueva_distancia = distancia_actual + distancia

            if nueva_distancia < distancias[vecino]:
                distancias[vecino] = nueva_distancia

                heapq.heappush(
                    frontera,
                    (nueva_distancia, vecino)
                )

    return distancias
```

```python
def a_star_graph_search(problem):
    nodo_inicial = Node(problem.initial)

    frontera = []
    contador = 0

    f_inicial = nodo_inicial.path_cost + problem.h(nodo_inicial.state)

    heapq.heappush(
        frontera,
        (f_inicial, contador, nodo_inicial)
    )

    explorados = set()

    while frontera:
        f_actual, _, nodo = heapq.heappop(frontera)

        if problem.is_goal(nodo.state):
            return nodo

        if nodo.state in explorados:
            continue

        explorados.add(nodo.state)

        for hijo in nodo.expand(problem):

            if hijo.state not in explorados:
                contador += 1

                f_hijo = hijo.path_cost + problem.h(hijo.state)

                heapq.heappush(
                    frontera,
                    (f_hijo, contador, hijo)
                )

    return None
```

```python
resultado = a_star_graph_search(problema)
```
