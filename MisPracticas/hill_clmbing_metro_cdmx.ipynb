```python
def crear_heuristica(grafo, objetivo):
    distancias = {estacion: float("inf") for estacion in grafo}
    distancias[objetivo] = 0

    cola = [(0, objetivo)]

    while cola:
        distancia_actual, estacion_actual = heapq.heappop(cola)

        if distancia_actual > distancias[estacion_actual]:
            continue

        for vecino, distancia in grafo[estacion_actual].items():
            nueva_distancia = distancia_actual + distancia

            if nueva_distancia < distancias[vecino]:
                distancias[vecino] = nueva_distancia
                heapq.heappush(cola, (nueva_distancia, vecino))

    return distancias
```

```python
def hill_climbing(grafo, inicio, objetivo, heuristica):
    actual = inicio
    camino = [actual]
    distancia_total = 0

    while actual != objetivo:

        vecinos = grafo[actual]

        mejor_vecino = min(
            vecinos,
            key=lambda vecino: heuristica[vecino]
        )

        if heuristica[mejor_vecino] >= heuristica[actual]:
            return None, distancia_total

        distancia_total += vecinos[mejor_vecino]
        actual = mejor_vecino
        camino.append(actual)

    return camino, distancia_total
```

```python
inicio = "Cuatro Caminos"
objetivo = "Pantitlán"

heuristica = crear_heuristica(metro_cdmx, objetivo)

camino, distancia = hill_climbing(
    metro_cdmx,
    inicio,
    objetivo,
    heuristica
)

print("Inicio:", inicio)
print("Objetivo:", objetivo)

if camino:
    print("\nCamino encontrado:")
    print(" -> ".join(camino))

    print("\nDistancia total:", distancia, "metros")
else:
    print("\nHill Climbing no pudo llegar al objetivo.")
```
